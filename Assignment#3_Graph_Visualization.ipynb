```python
import networkx as nx
import matplotlib.pyplot as plt
import os
```

```python
data = nx.read_edgelist('/Users/frazaslam/Desktop/DATA 620 - Web Analytics/reptilia-tortoise-network-sl.edges', nodetype=int, data=[('year', int)])
```

```python
print(data.number_of_edges())
```

```python
print(data.number_of_nodes())
```

```python
nx.draw(data, with_labels=True)
plt.show()
```

```python
nx.is_connected(data)
```

```python
nx.number_connected_components(data)
```

```python
## Diameter For the Largest Group 
## Because there are 2 separate groups we measure the diameter in the largest group
largest = max(nx.connected_components(data), key=len)
data_big = data.subgraph(largest)

nx.diameter(data_big)
```

```python
## This Details the shortest path between 10 and 3 which is also the Diameter representation of the largest group
nx.shortest_path(data_big, 10, 3)
```

```python
## Degree -- How many other torotises one tortoise shared burrows with

degrees = dict(data.degree())

degrees
```

```python
## Node 8 shared the most burrows with other tortoises, which is represented by 4 edges or connections
print(sorted(degrees.items(), key=lambda x: x[1], reverse=True))
```
